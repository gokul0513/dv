```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[2,4,6,8,10]
plt.plot(x,y)
plt.show()
```

```python
x=[1,2,3,4,5]
y=[2,4,6,8,10]
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.xlim(1,10)
plt.ylim(5,15)
plt.title("Line plot Example")
plt.plot(x,y)
plt.savefig("sample.png")
plt.show()
```

```python
x=[101,102,103,104,105]
y=[55,80,90,77,63]
plt.xlabel("roll no")
plt.ylabel("percentage")
plt.xlim(100,106)
plt.ylim(50,95)
plt.title("Student marks")
plt.plot(x,y,color='red',linestyle='dashdot',linewidth=5,marker='*',markerfacecolor='blue',markersize=15)
plt.show()
```

```python
x=[1,2,3,4,5]
y=[2,4,6,8,10]
x1=[1,2,3,4]
y1=[5,6,9,2]
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.xlim(1,10)
plt.ylim(5,15)
plt.title("Line plot Example")
plt.plot(x,y,label="line1",color='red',linestyle='dotted',linewidth=5,marker='*',markerfacecolor='blue',markersize=15)
plt.plot(x1,y1,label="line2",alpha=0.2)
plt.legend()
plt.savefig("sample.png")
plt.show()
```

```python
import matplotlib.pyplot as plt
stud=['A','B','C','D','E']
marks=[56,60,45,70,82]
plt.bar(stud,marks,color="black",width=0.7,edgecolor="red",label="marks")
plt.legend()
plt.title("student marks")
plt.xlabel("stud")
plt.ylabel("marks")
plt.show()
```
