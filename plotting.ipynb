```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[2,4,6,8,10]
plt.xlabel("x axis")
plt.ylabel("y-axis")
plt.title("line plot example")
plt.xlim(0,10)
plt.ylim(0,15)
plt.plot(x,y)
plt.savefig("sample.png")
plt.show()
```

```python
x=[1,2,3,4,5]
y=[88,80,62,98,100]
plt.xlabel("roll number")
plt.ylabel("marks")
plt.title("student marks")
plt.xlim(0,6)
plt.ylim(60,120)
plt.plot(x,y,color="green",linestyle="dotted",linewidth=3,marker=".",markerfacecolor="blue",markersize=20)
plt.show()
```

```python
x=[1,2,3,4,5]
y=[2,4,6,8,10]
x1=[1,2,3,4,5]
y1=[3,2,8,5,1]
plt.xlabel("x axis")
plt.ylabel("y-axis")
plt.title("line plot example")
plt.xlim(0,10)
plt.ylim(0,12)
plt.plot(x,y,label="line 1",color="green",linestyle="dotted",linewidth=3,marker=".",markerfacecolor="blue",markersize=20)
plt.plot(x1,y1,label="line 2",color="yellow",linestyle="solid",linewidth=3,marker=".",markerfacecolor="red",markersize=20)
plt.legend()
plt.savefig("sample2.png")
plt.show()
```
